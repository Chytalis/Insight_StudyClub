```python
df_raw = pd.read_csv("SampleSuperstore.csv")

print("Ukuran data (baris, kolom):", df_raw.shape)
display(df_raw.head())
display(df_raw.tail())
```

```python
print("Missing value per kolom:")
print(df_raw.isna().sum())
print("\nJumlah baris duplikat:", df_raw.duplicated().sum())

print("\nJumlah nilai unik per kolom:")
print(df_raw.nunique())

for col in ["Ship Mode", "Segment", "Region", "Category"]:
    print(f"\n{col}:")
    print(df_raw[col].value_counts())
```

```python
df = df_raw.copy()
```

```python
# 1. Rapikan nama kolom & spasi pada teks
df.columns = df.columns.str.strip()
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].str.strip()
```

```python
# 2. Hapus duplikat
before = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print(f"Duplikat dihapus: {before - len(df)} baris")
```

```python
# 3. Perbaiki tipe data (Postal Code = kode, bukan angka hitung)
df["Postal Code"] = df["Postal Code"].astype(str).str.zfill(5)
for col in ["Ship Mode", "Segment", "Region", "Category", "Sub-Category"]:
    df[col] = df[col].astype("category")
```

```python
# 4. Validasi nilai
print("Sales <= 0       :", (df["Sales"] <= 0).sum())
print("Quantity <= 0    :", (df["Quantity"] <= 0).sum())
print("Discount di luar 0-1:", (~df["Discount"].between(0, 1)).sum())
```

```python
# 5. Feature engineering
df["Profit Margin (%)"] = df["Profit"] / df["Sales"] * 100
df["Is Loss"] = df["Profit"] < 0
df["Discount Band"] = pd.cut(
    df["Discount"],
    bins=[-0.001, 0, 0.2, 0.4, 1.0],
    labels=["0%", "1-20%", "21-40%", ">40%"],
)
```

```python
# 6. Deteksi outlier (IQR) -> hanya ditandai, tidak dihapus
def count_outliers(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum()

for col in ["Sales", "Quantity", "Profit"]:
    print(f"Outlier {col}: {count_outliers(df[col])}")
```

```python
# 7. Simpan hasil cleaning
df.to_csv("superstore_clean.csv", index=False)
print("\nShape setelah cleaning:", df.shape)
display(df.head())
```
